```python
#now lets create a data frame for the summary
with open("t20_wc_match_results.json") as f:
    data=json.load(f)
df_match= pd.DataFrame(data[0]["matchSummary"])
df_match
```

```python
#check how many elements it has---- we can do this by using the method called "shape"
df_match.shape
```

```python
#use score card as a key for the above particular data frame because to identify uniquely, treat it as match id because we can connect with other tables
```

```python
#rename the scorecard column
```

```python
df_match.rename({"scorecard":"match_id"},axis=1, inplace= True)   #we have to use inplace=true otherwise it creates a new dataframe/table
df_match.head() #to access top 5 rows
```

```python
#create dictionary to store the items or elements
match_ids_dict={}
for index,row in df_match.iterrows():
    key1=row["team1"]+ " Vs " +row["team2"]
    key2=row["team2"]+" Vs "+row["team1"]
    match_ids_dict[key1]=row["match_id"]
    match_ids_dict[key2]=row["match_id"]
match_ids_dict
```

```python
df_match.to_csv('dim_match_summary.csv', index = False)
```

**Batting Summary**

```python
#open batting summary json file
with open("t20_wc_batting_summary.json") as f:
    data=json.load(f)
data
```

```python
# here we have details of all matches to get the summary of one match we have to execute the following code
with open("t20_wc_batting_summary.json") as f:
    data=json.load(f)

    for rec in data:
        rec["battingSummary"]
data
```

```python
#append ever thing into one array that is merge all the matches
#to create a single list in pandas we use the code
with open("t20_wc_batting_summary.json") as f:
    data=json.load(f)

    all_records=[]
         
    for rec in data:
        all_records.extend(rec["battingSummary"])
        #extend is used when we have one array and we are appending another array
all_records
```

```python
#now create a data frame with the give data
with open("t20_wc_batting_summary.json") as f:
    data=json.load(f)

    all_records=[]
         
    for rec in data:
        all_records.extend(rec["battingSummary"])
df_batting=pd.DataFrame(all_records)
df_batting.head()
         
 
```

```python
df_batting.head(11)
```

```python
# now w have to convert the dismissal column to out ir not out column if the box is empty then the player is not out and if the box have some thing the player out.
df_batting["out/not_out"]= df_batting.dismissal.apply(lambda x:"out" if len(x)>0 else "not_out")
df_batting.head()
```

```python
df_batting.head(11)
```

```python
# now there is no need of dismissal column so drop that column
df_batting.drop(columns=["dismissal"],inplace=True)
df_batting.head(11)
```

```python
# now if you observe there are some issue with batsman name as it contains the special characters
df_batting["batsmanName"]=df_batting["batsmanName"].apply(lambda x: x.replace("â€",""))
df_batting["batsmanName"]=df_batting["batsmanName"].apply(lambda x: x.replace("/xa0",""))
df_batting.head(11)
```

```python
# now the main task is to connect this particular data frame to the previously modified match data frame in both we have match column same so we have to merge ,
```

```python
# now lts check the match_id for a particular match
match_ids_dict["Namibia Vs Sri Lanka"]
```

```python

df_batting["match_id"]= df_batting["match"].map(match_ids_dict)
df_batting.head()
#now we got the match_id's for all the batting table also
```

```python
#now we have to export this particular file into CSV file
df_batting.to_csv("fact_batting_summary.csv", index = False)
```

```python
# now we have to process the bowling summary
import json
with open('t20_wc_bowling_summary.json') as f:
    data = json.load(f)
    all_records = []
    for rec in data:
        all_records.extend(rec['bowlingSummary'])
all_records[:2]
```

```python
df_bowling = pd.DataFrame(all_records)
print(df_bowling.shape) #shape of the table
df_bowling.head()  # returns the top 5 elements
```

```python
df_bowling['match_id'] = df_bowling['match'].map(match_ids_dict)   # here we are adding the elements with the match ids
df_bowling.head()
```

```python
df_bowling.to_csv('fact_bowling_summary.csv', index = False)   # download the csv file of the data
```

```python
with open('t20_wc_player_info.json') as f:
    data = json.load(f)
```

```python
df_players = pd.DataFrame(data)

print(df_players.shape)
df_players.head(10)
```

```python
# we can observe some weird characters over the names of the players we have to clean them
```

```python
df_players['name'] = df_players['name'].apply(lambda x: x.replace('â€', ''))
df_players['name'] = df_players['name'].apply(lambda x: x.replace('†', ''))
df_players['name'] = df_players['name'].apply(lambda x: x.replace('\xa0', ''))
df_players.head(10)
```
